```python
# Open the tracking data file in read mode
with open('tactical/BAR-ATH_tactical_atd.txt', 'r') as file:
    line_reports = []
    all_counts_match = True
    first_line_element_count = None
    
    # Process up to the first 10 lines
    for i in range(10):
        line = file.readline()
        
        # Stop early if the file has fewer than 10 lines
        if not line:
            break
            
        clean_line = line.strip()
        if not clean_line:
            continue  # Skip blank lines if any
            
        # 1. Split the Main Key from the Values using the first colon
        key, all_values = clean_line.split(':', 1)
        
        # 2. Split into groups by semicolon (ignoring empty trailing semicolons)
        groups = [g for g in all_values.split(';') if g]
        
        # 3. Count individual elements inside all groups combined
        total_elements = 0
        for group in groups:
            individual_values = group.split(',')
            total_elements += len(individual_values)
            
        # Establish the baseline count from the first line
        if first_line_element_count is None:
            first_line_element_count = total_elements
            
        # Check if this line matches the baseline
        if total_elements != first_line_element_count:
            all_counts_match = False
            
        # Store metadata for reporting
        line_reports.append({
            'line_num': i + 1,
            'key': key,
            'group_count': len(groups),
            'element_count': total_elements
        })

# --- Final Report ---
if all_counts_match and line_reports:
    print(f"✅ Success! Structure is perfectly consistent.")
    print(f"All checked lines contain exactly {first_line_element_count} total individual elements./n")
else:
    print("❌ Mismatch found! The number of elements per line varies./n")

# Print the breakdown for visual confirmation
print(f"{'Line':<6} | {'Key':<12} | {'Semicolon Groups':<16} | {'Total Elements':<14}")
print("-" * 58)
for row in line_reports:
    print(f"{row['line_num']:<6} | {row['key']:<12} | {row['group_count']:<16} | {row['element_count']:<14}")
```
